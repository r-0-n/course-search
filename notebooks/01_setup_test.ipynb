```python
!pip install -U sentence-transformers transformers accelerate

from sentence_transformers import SentenceTransformer

model = SentenceTransformer("google/embeddinggemma-2")
print("Loaded OK")


query ="How to fix a puncture tyre?"
docs  = [
    "Suruh adnin kasi bagus",
    "Jack the car, unfast the tyre lug nuts, pull the puncture tyre and replace it with a spare tyre",
    "I just want to drive, man",
    "Wing it,",
    "Switch to airplane mode",
    "Here a step on how to make pasta carbonara",
    "The missle know where it is where it doesn't",
    "Only when the star allign, Ferrari can get first podium finish",
    "Oye, Tony, arréglalo.",
]

q = model.encode(query, prompt_name="SearchQuery")
d = model.encode(docs, prompt_name="Document")
scores = model.similarity(q, d)[0]

for doc, s in sorted(zip(docs, scores), key=lambda x: -x[1]):
  print(round(float(s), 3), doc)
```
